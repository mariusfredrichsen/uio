## 1. Create a Virtual Environment (`venv`)

Instead of installing libraries globally (which can break your system), we will create a folder named `.venv` that contains a fresh Python installation just for this course.

### Step 1: Open your Terminal
* **VS Code:** `Terminal` -> `New Terminal`
* **Windows:** PowerShell or Command Prompt
* **Mac/Linux:** Terminal

### Step 2: Run the Creation Command
Type the following command in your terminal (make sure you are in your project folder!):

```bash
# Windows/Mac/Linux
python -m venv .venv
```
*(Note for Mac/Linux: If `python` doesn't work, try `python3`)*

**What happens?** A new folder named `.venv` will appear in your file explorer. It contains a clean Python setup.

## 2. Activate the Environment

You need to "enter" this environment before installing anything.

### Windows:
```powershell
.venv\Scripts\activate
```

### Mac / Linux:
```bash
source .venv/bin/activate
```

**Success Check:** You should see `(.venv)` appear at the beginning of your terminal line.

## 3. Select Kernel in VS Code (Important!)

If you are running this notebook in VS Code:

## 6. Final Verification
Run this code to confirm everything is working correctly.

````python
import torch
import sys

print(f"🐍 Python Path: {sys.executable}")
print(f"🔥 PyTorch Version: {torch.__version__}")

print("\n--- GPU Capability Check ---")

if torch.cuda.is_available():
    gpu_name = torch.cuda.get_device_name(0)
    cuda_ver = torch.version.cuda
    
    try:
        cc_major, cc_minor = torch.cuda.get_device_capability(0)
        print(f"✅ CUDA is available!")
        print(f"   GPU Name:       {gpu_name}")
        print(f"   Compute Cap:    {cc_major}.{cc_minor}")
        print(f"   PyTorch CUDA:   {cuda_ver}")
        print(f"   Total GPUs:     {torch.cuda.device_count()}")
        for i in range(torch.cuda.device_count()):
            print(f"        Device {i}: {torch.cuda.get_device_name(i)}")

        # Logic Check for RTX 50-series (Blackwell Architecture >= 10.0)
        is_blackwell = (cc_major >= 10)
        is_old_cuda = (float(cuda_ver.split('.')[0]) < 13)

        if is_blackwell and is_old_cuda:
             print("\n❌ CRITICAL WARNING: CONFIGURATION MISMATCH")
             print(f"   You have a new Blackwell GPU ({gpu_name}) but PyTorch is using CUDA {cuda_ver}.")
             print("   This will crash. Please run OPTION A above (CUDA 13) to fix this.")
        else:
             print("\n🎉 Configuration looks good! You are ready for the course.")
             
    except Exception as e:
        print(f"✅ CUDA available, but couldn't read capability: {e}")

elif torch.backends.mps.is_available():
    print("✅ Apple Metal (MPS) is available (Mac M1/M2/M3).")
    print("🎉 You are ready for the course.")
else:
    print("⚠️ No GPU detected. You are running on CPU mode.")
    print("   This is fine! We will start with small models where CPU is sufficient.")
````
